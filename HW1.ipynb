Введем функцию преобразования string to integer с удалением лишних пробелов:

````python
def parse_int(s: str) -> int | None:
  try:
    return int(s.strip())
  except ValueError:
    return None
````

Введем тестовые значения:

````python
nums_part1 = ["12", " -7 ", "12a", ""]
````

Применим функции к набору тестовых значений и выведем результат в следующем формате:
```
{'исходное значение'} тип данных исходного значения | {итоговое значение} тип данных итогового значения
```

````python
for value in nums_part1:
  print("from:", {value}, " ", type(value), "| to:", {parse_int(value)}," ",type(parse_int(value)) )
````

#Часть 2

Реализуем функции:

1.  Подсчитаем количество значений >0 в заданном наборе
2.  Подсчитаем суммы всех чётных чисел в наборе
3.  Выявим первое число, кратное задонному (k) c прерыванием цикла

````python
def count_positive(nums_part2: list[int]) -> int:
  k = 0
  for num in nums_part2:
    if num > 0:
      k += 1
  return k


def sum_even(nums_part2: list[int]) -> int:
  sum = 0
  for num in nums_part2:
    if num % 2 == 0:
      sum += num
  return sum


def first_divisible(nums_part2: list[int], k: int) -> int | None:
  res = None

  for num in nums_part2:
    if num % k == 0:
      res = num
      break
  return res
````

Введем тестовый набор значений для второй части. Выведем результаты:

````python
nums_part2 = [12, -5, 7, -11, 18,
        4, -2, 9, 15, -8, 22, 3,
        -1, 6, 14, -9, 5, 8, 11,
        -13, 20, 1, -4, 16]

print("Количество положительных чисел:", count_positive(nums_part2))
print("Сумма чётных чисел:", sum_even(nums_part2))
print("Первое число, которое делится на 5:", first_divisible(nums_part2, 5))
print("Первое число, которое делится на 13:", first_divisible(nums_part2, 13))
print("Первое число, которое делится на 100:", first_divisible(nums_part2, 100))
````

# Часть 3

Введем функцию сортировки списка словарей по заданному ключу с проверкой наличия ключа:

````python
def sort(items: list[dict], sort_by: str) -> list[dict] | None:
    try:
        if sort_by == "rating":
            return sorted(items, key=lambda x: x[sort_by], reverse=True)
        else:
            return sorted(items, key=lambda x: x[sort_by])
    except KeyError:
        return None
````

````python
items = [
  {"name": "A", "price": 120, "rating": 4.7},
  {"name": "B", "price": 90, "rating": 4.2},
  {"name": "C", "price": 150, "rating": 4.9},
]
````

В

````python
print(sort(items, "price"))
print(sort(items, "rating"))
print(sort(items, "abcd"))
````

# Часть 4

Введем класс Chocolate с параметрами name и price, методом расчета скидочной стоимости и конструктором с проверкой цены на неотрицательность:

````python
class Chocolate:
  name: str
  price: float

  def __init__(self, name: str, price: float):
        if price <= 0:
            raise ValueError("Цена не может быть меньше 0")
        self.name = name
        self.price = price

  def discount(self, percent: int) -> float:
    return self.price * (1 - percent / 100)
````

Создадим два объекта класса с заданными параметрами name и price и рассчитаем скидочную стоимость:

````python
chocolate1 = Chocolate("ch1", 120.0)
chocolate2 = Chocolate("ch2", 130.0)

print(chocolate1.name, chocolate1.discount(10))
print(chocolate2.name, chocolate2.discount(13))
````
