Найдем наиболее частотные имена среди женщин и мужчин в датафрейме.

Новым столбцом в датафрейме добавим ```firstname```, для каждой строки вызовем функцию ```real_first_name```

Поскольку женские имена иногда указываются в скобках рядом с мужским, для женщин необходима проверка на наличие скобок в имени. Для мужчин и женщин без скобок в имени используем первое слово после точки.

Далее группируем по полу и имени и выводим id (имя) первого вхождения максимального значения (```value_counts``` по имени)

````python
def real_first_name(name):
    if '(' in name:
        inner = name.split('(')[1].split(')')[0]
        return inner.split()[0].replace('"', '')

    if '.' in name:
        parts = name.split('.')
        first_name = parts[1].strip().split(' ')[0]
        return first_name.replace('"', '')

    return name

df['firstname'] = df['Name'].apply(real_first_name)

print('Самое популярное мужское имя:', df[df['Sex'] == 'male']['firstname'].value_counts().idxmax()
)
print('Самое популярное женское имя:', df[df['Sex'] == 'female']['firstname'].value_counts().idxmax()
)
````

Группируем данные по полу, классу и имени. С помощью функции ```mode``` найдем самые частотные имена по классам - для женщин и мужчин отдельно:

````python
df.groupby(['Pclass', 'Sex'])['firstname'].agg(lambda x: x.mode())
````
